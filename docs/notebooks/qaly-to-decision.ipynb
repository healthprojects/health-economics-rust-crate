# From QALYs to a funding decision

This notebook composes three small, spec-backed calculations. Values are illustrative.

```rust
:dep health-economics = { path = "../.." }
```

```rust
use health_economics::quality_adjusted_life_year as qaly;
use health_economics::incremental_cost_effectiveness_ratio as icer;
use health_economics::net_monetary_benefit as nmb;
```

A service removes a six-month wait for 400 people, changing utility from 0.60 to 0.85.

```rust
let per_patient = qaly::qaly_loss_from_delay(0.5, 0.6, 0.85);
let total_qalys = qaly::population_qalys(per_patient, 400.0);
assert!((total_qalys - 50.0).abs() < 1e-9);
total_qalys
```

At an incremental annual cost of £500,000, compare cost per QALY and net monetary benefit at a £20,000 threshold.

```rust
let incremental_cost = 500_000.0;
let threshold = 20_000.0;
let cost_per_qaly = icer::icer(incremental_cost, total_qalys).unwrap();
let net_benefit = nmb::net_monetary_benefit(total_qalys, incremental_cost, threshold);
assert!((cost_per_qaly - 10_000.0).abs() < 1e-9);
assert!((net_benefit - 500_000.0).abs() < 1e-9);
(cost_per_qaly, net_benefit, nmb::adopt(net_benefit))
```

The two rules agree: the ICER is below the threshold and NMB is positive. Before using this result, vary uptake, cost, utility, and displacement assumptions using the uncertainty tutorial.
